```python
## Datasets Used

1. orders.csv  
   - Contains transactional order data (order_id, user_id, restaurant_id, order_date, total_amount)

2. users.json  
   - Contains user details (user_id, name, city, membership)

3. restaurants.sql  
   - Contains restaurant details (restaurant_id, restaurant_name, cuisine, rating)
```

```python
## Data Integration

All datasets were merged using LEFT JOIN to retain all order records.
```

```python
-- Final merged dataset with rating included
CREATE TABLE final_food_delivery_dataset_v2 AS
SELECT
    o.order_id,
    o.order_date,
    o.total_amount,
    u.user_id,
    u.name AS user_name,
    u.city AS user_city,
    u.membership,
    r.restaurant_id,
    r.restaurant_name,
    r.cuisine,
    r.rating
FROM orders o
LEFT JOIN users u ON o.user_id = u.user_id
LEFT JOIN restaurants r ON o.restaurant_id = r.restaurant_id;
```

```python
-- Q1: City with highest revenue from Gold members
SELECT user_city, SUM(total_amount) AS total_revenue
FROM final_food_delivery_dataset_v2
WHERE membership = 'Gold'
GROUP BY user_city
ORDER BY total_revenue DESC;
```

```python
-- Q2: Cuisine with highest average order value
SELECT cuisine, AVG(total_amount) AS avg_order_value
FROM final_food_delivery_dataset_v2
GROUP BY cuisine
ORDER BY avg_order_value DESC;
```

```python
-- Q3: Users who spent more than 1000
SELECT COUNT(*) FROM (
    SELECT user_id
    FROM final_food_delivery_dataset_v2
    GROUP BY user_id
    HAVING SUM(total_amount) > 1000
) t;
```

```python
-- Q4: Rating range with highest revenue
SELECT
  CASE
    WHEN rating BETWEEN 3.0 AND 3.5 THEN '3.0–3.5'
    WHEN rating BETWEEN 3.6 AND 4.0 THEN '3.6–4.0'
    WHEN rating BETWEEN 4.1 AND 4.5 THEN '4.1–4.5'
    WHEN rating BETWEEN 4.6 AND 5.0 THEN '4.6–5.0'
  END AS rating_range,
  SUM(total_amount) AS total_revenue
FROM final_food_delivery_dataset_v2
GROUP BY rating_range
ORDER BY total_revenue DESC;
```

```python
-- Q5: Cuisine with lowest restaurant count but high revenue
SELECT cuisine,
       COUNT(DISTINCT restaurant_id) AS restaurant_count,
       SUM(total_amount) AS total_revenue
FROM final_food_delivery_dataset_v2
GROUP BY cuisine
ORDER BY restaurant_count ASC, total_revenue DESC;
```

```python
-- Q6: Membership + cuisine combination with highest revenue
SELECT membership, cuisine, SUM(total_amount) AS total_revenue
FROM final_food_delivery_dataset_v2
GROUP BY membership, cuisine
ORDER BY total_revenue DESC;
```

```python
-- Q7: Quarter with highest revenue
SELECT
  CONCAT('Q', QUARTER(STR_TO_DATE(order_date, '%Y-%m-%d'))) AS quarter,
  SUM(total_amount) AS total_revenue
FROM final_food_delivery_dataset_v2
GROUP BY quarter
ORDER BY total_revenue DESC;
```

```python
-- Total orders by Gold members
SELECT COUNT(*) FROM final_food_delivery_dataset_v2
WHERE membership = 'Gold';
```

```python
-- Revenue from Hyderabad
SELECT ROUND(SUM(total_amount)) FROM final_food_delivery_dataset_v2
WHERE user_city = 'Hyderabad';
```

```python
-- Distinct users
SELECT COUNT(DISTINCT user_id) FROM final_food_delivery_dataset_v2;
```

```python
-- Average order value for Gold members
SELECT ROUND(AVG(total_amount),2)
FROM final_food_delivery_dataset_v2
WHERE membership = 'Gold';
```

```python
-- Orders in top Gold-revenue city (Chennai)
SELECT COUNT(*) FROM final_food_delivery_dataset_v2
WHERE membership = 'Gold' AND user_city = 'Chennai';
```
